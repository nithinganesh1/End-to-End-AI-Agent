### Messages

Messages are structured units used to represent communication and conversation context between humans and AI models.

- **Role** – Identifies who created the message, such as `user` (human) or `assistant` (AI).
- **Content** – Contains the actual text, instructions, questions, or responses.
- **Metadata** – Stores additional information such as timestamps, message IDs, or other contextual details.

**Example:**

- Human gives an input → **Human/User Message**
- AI model generates an output → **AI/Assistant Message**

### Text Prompt

Text prompts are simple text strings used to provide instructions or questions to an AI model. They are ideal when there is no need to store or maintain conversation history.

* No conversation history is required.
* Best for simple tasks.
* Suitable for standalone requests.

**Example:**
`"Please Explailn what is messages in llm."`


### Message Prompt

Message prompts allow us to pass a list of structured message objects to an AI model. This format is useful for maintaining conversation history and managing different types of interactions.

**Message Types:**

* **System Message** – Provides high-level instructions that define the AI's behavior, role, rules, and response style.

* **Human Message** – Represents the input, questions, or instructions provided by the human user to the AI.

* **AI Message** – Represents the responses or outputs generated by the AI model during the conversation.

* **Tool Message** – Contains results or information returned by external tools, such as databases, APIs, search engines, or calculators.


```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
messages = [
    SystemMessage("You are a teacher Explain thigs as samll story example"),
    HumanMessage("what is type of messages in llm")
]
```

```python
response = model.invoke(messages)
```

```python
print(response.content)
```

```python
message = HumanMessage(
    content="How do I create a REST API?",
    metadata={
        "user_id": "123",
        "timestamp": "2026-09-01",
        "language": "English"
    }
)

print(message.content)
print(message.metadata)
```

```python
messages = [
    HumanMessage(content="How do I create a REST API?"),
    AIMessage(content="You can create a REST API using FastAPI."),
    HumanMessage(content="Can you show me a simple example?")
]

response = model.invoke(messages)

print(response.content)
```

```python
from langchain_core.messages import ToolMessage

def get_weather(location):
    return f"The weather in {location} is 28°C and cloudy."

messages = [
    HumanMessage(content="What is the weather in Kerala?"),
    
    AIMessage(
        content="",
        tool_calls=[{
            "name": "get_weather",
            "args": {"location": "Kerala"},
            "id": "call_123"
        }]
    ),
    
    ToolMessage(
        content=get_weather("Kerala"),
        tool_call_id="call_123"
    )
]

for message in messages:
    print(message)
```
