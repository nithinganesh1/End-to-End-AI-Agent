### Langchain

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['GOOGLE_API_KEY']=os.getenv('GOOGLE_API_KEY')
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
)
```

```python
#basic agent 1
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[],
    system_prompt="you are an helpul assistant"
)
agent
```

```python
#basic agent 2
def get_wether(city:str)-> str:
    """get the wether for the city"""
    return f"wether for the {city} is rainy"

agent = create_agent(
    model=llm,
    tools=[get_wether],
    system_prompt="you are an helpul assistant"
)
agent
```

```python
### run the agent
agent.invoke({"messages":[{"role":"user","content":"what is the wether in kozhikode"}]})
```

```python
### run the agent
response=agent.invoke({"messages":"what is the wether in kozhikode"})
response['messages']
```
