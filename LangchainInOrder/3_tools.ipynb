## Tools

**Tools** are functions or services that an AI uses to **perform tasks** or **access information beyond its own knowledge**.

**Simple Definition:**
> Tools allow an AI to interact with the outside world to complete tasks.

### Examples
- 🌦️ Get the current weather
- 🌐 Search the internet
- 🧮 Perform calculations
- 📧 Send an email
- 📅 Create a calendar event
- 🗄️ Query a database

### Why Use Tools?
Without tools, an AI can only answer based on its existing knowledge. With tools, it can perform real actions and access up-to-date information.

```python
#model=init_chat_model("groq:llama-3.1-8b-instant")
model=init_chat_model("google_genai:gemini-2.5-flash")
response=model.invoke("why cars have breaks?")
response.content
```

```python
from langchain.tools import tool #use as a decrator

@tool
def get_wether(location:str)->str:
    """get wether of a location"""      # when llm bind with this tool llm will able to identify
    return f"The weather in {location} is rainy. Temperature: 27°C. Humidity: 80%."   # the funtionality from the discription

model_with_tools = model.bind_tools([get_wether])
```

```python
response = model_with_tools.invoke("what is the wether in kerala?")
```

```python
for tool_call in response.tool_calls:
    # view tool call made by the model
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
```

### Tool Exicution Loop

```python
messages = [
    {
        "role": "user",
        "content": "what is the weather in kasaragod?"
    }
]

# Step 1: Model generates tool call
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

# Step 2: Execute tool
for tool_call in ai_msg.tool_calls:
    tool_result = get_wether.invoke(tool_call)
    messages.append(tool_result)

# Step 3: Generate final response
final_result = model.invoke(messages)
#print(messages)
print(final_result.text)
```
